```python
import pandas as pd

population = pd.Series(
    [20.7, 19.1, 8.4, 7.1, 4.5],
    index=['Mumbai', 'Delhi', 'Bengaluru', 'Chennai', 'Kolkata']
)

print(population['Bengaluru'])
print(population['Delhi':'Chennai'])
```

```python
import pandas as pd

prices = {
    'Laptop': 55000,
    'Mobile': 25000,
    'Tablet': 18000
}

s = pd.Series(prices)

print(s)
print(s.index)
print(s.index.equals(pd.Index(prices.keys())))
```

```python
import pandas as pd

area = pd.Series(
    [13060, 191791, 308252, 196024, 114223],
    index=['Kerala', 'Karnataka', 'Rajasthan', 'Gujarat', 'Tamil Nadu']
)

population = pd.Series(
    [3.34, 6.11, 8.21, 6.04, 7.21],
    index=['Kerala', 'Karnataka', 'Rajasthan', 'Gujarat', 'Tamil Nadu']
)

states = pd.DataFrame({'area': area, 'population': population})
states['density'] = states['population'] / states['area']

print(states)
```

```python
import pandas as pd

indA = pd.Index([2, 3, 5, 7, 11])
indB = pd.Index([1, 3, 5, 7, 9])

print("Intersection:", indA.intersection(indB))
print("Union:", indA.union(indB))
print("Symmetric Difference:", indA.symmetric_difference(indB))
```
